```python
from scipy.stats import zscore

# Columns for Z-score calculation
z_score_cols = ['T2M', 'T2M_MAX', 'T2M_MIN', 'PRECTOTCORR', 'RH2M', 'WS2M', 'WS2M_MAX']

# Calculate Z-scores for the selected columns
for col in z_score_cols:
    df[f'{col}_Zscore'] = zscore(df[col], nan_policy='omit')

# Identify outliers where |Z| > 3
outlier_mask = pd.DataFrame()
for col in z_score_cols:
    outlier_mask = pd.concat([outlier_mask, (df[f'{col}_Zscore'].abs() > 3)], axis=1)

# A row is an outlier if any of the specified columns has an absolute Z-score > 3
df['is_outlier'] = outlier_mask.any(axis=1)

# Report the count of outliers
outlier_count = df['is_outlier'].sum()
print(f"Number of rows with at least one outlier (Z-score > 3): {outlier_count}")

display(df[df['is_outlier']].head())
```

## Outlier Treatment Strategy
**Decision:** Retain all identified outliers.

**Reasoning:**
* **Scientific Relevance:** In climate datasets, outliers often represent genuine extreme weather events (e.g., peak temperatures or heavy precipitation). Removing them would result in a loss of critical information regarding climate volatility.
* **Data Integrity:** There is no evidence to suggest these points are the result of sensor errors or manual data entry mistakes.
* **Modeling Impact:** Retaining these points ensures the model accounts for "tail risks" and accurately reflects the real-world variance of the Ethiopian climate.

```python
# 1. Drop rows with missing values in more than 30% of the columns
limit = int(len(df.columns) * 0.7)
df = df.dropna(thresh=limit)

# 2. Apply forward-fill to the remaining missing weather variables
df = df.ffill()
```

## Handling Missing Values
**Strategy:** Hybrid approach using row-wise deletion and Forward-Fill (ffill).

**1. Row Deletion (30% Threshold):**
Rows missing more than 30% of their features were removed. When a significant portion of
a record is missing, imputation becomes unreliable and may introduce synthetic noise
rather than capturing actual environmental trends.

**2. Forward-Fill (ffill) for Weather Variables:**
For the remaining missing values, I applied a forward-fill strategy.
* **Reasoning:** Climate data is chronological and highly autocorrelated. Weather
conditions at time $t$ are typically the most accurate predictors for time $t+1$.
Forward-filling preserves this temporal continuity without shifting the mean
as drastically as mean-imputation might.

# Time series Analysis

```python
import matplotlib.pyplot as plt


# Create a datetime index
if 'YEAR' in df.columns and 'MO' in df.columns:
    df['Date'] = pd.to_datetime(df[['YEAR', 'MO']].assign(DAY=1))
else:
    # Adjust this if your date column has a different name
    df['Date'] = pd.to_datetime(df['Date'])

df.set_index('Date', inplace=True)

# 2. Resample to Monthly Average
monthly_t2m = df['T2M'].resample('M').mean()

# 3. Identify Warmest and Coolest Months
warmest_month = monthly_t2m.idxmax()
warmest_val = monthly_t2m.max()

coolest_month = monthly_t2m.idxmin()
coolest_val = monthly_t2m.min()

# 4. Plotting
plt.figure(figsize=(14, 6))
plt.plot(monthly_t2m.index, monthly_t2m.values, color='teal', linewidth=2, label='Monthly Avg T2M')

# Annotate Warmest
plt.annotate(f'Warmest: {warmest_val:.1f}°C\n({warmest_month.strftime("%b %Y")})',
             xy=(warmest_month, warmest_val),
             xytext=(warmest_month, warmest_val + 1),
             arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=8),
             horizontalalignment='center', color='red', fontweight='bold')

# Annotate Coolest
plt.annotate(f'Coolest: {coolest_val:.1f}°C\n({coolest_month.strftime("%b %Y")})',
             xy=(coolest_month, coolest_val),
             xytext=(coolest_month, coolest_val - 2),
             arrowprops=dict(facecolor='blue', shrink=0.05, width=1, headwidth=8),
             horizontalalignment='center', color='blue', fontweight='bold')

plt.title('Monthly Average Temperature (T2M) in Ethiopia (2015–2026)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Temperature (°C)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()

plt.savefig('monthly_t2m_plot.png')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Resample to Monthly Total Precipitation
monthly_prec = df['PRECTOTCORR'].resample('M').sum()

# Calculate average precipitation per month across all years
monthly_prec_avg = monthly_prec.groupby(monthly_prec.index.month).mean()

# Identify peak rainy season months (top 3 months by average precipitation)
peak_months = monthly_prec_avg.nlargest(3).index.tolist()

# Prepare data for plotting
plot_df = pd.DataFrame({'Month': monthly_prec_avg.index, 'Avg_Precipitation': monthly_prec_avg.values})
plot_df['Month_Name'] = plot_df['Month'].apply(lambda x: pd.to_datetime(str(x), format='%m').strftime('%b'))

# Sort by month name for proper display order
month_order = [pd.to_datetime(str(i), format='%m').strftime('%b') for i in range(1, 13)]
plot_df['Month_Name'] = pd.Categorical(plot_df['Month_Name'], categories=month_order, ordered=True)
plot_df = plot_df.sort_values('Month_Name')

# Plotting
plt.figure(figsize=(12, 6))
sns.barplot(x='Month_Name', y='Avg_Precipitation', data=plot_df, palette='viridis')

# Annotate peak rainy season months
for index, row in plot_df.iterrows():
    if row['Month'] in peak_months:
        plt.text(index, row['Avg_Precipitation'], f'{row['Avg_Precipitation']:.1f}mm',
                 color='red', ha='center', va='bottom', fontweight='bold')

plt.title('Average Monthly Total Precipitation (PRECTOTCORR) in Ethiopia (2015-2026)', fontsize=14)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Average Total Precipitation (mm)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Interpretation of Monthly Precipitation Trends

The bar chart illustrating the average monthly total precipitation reveals several key trends and patterns in Ethiopia's climate between 2015 and 2026:

*   **Clear Rainy Season:** There is a distinct rainy season, with precipitation significantly increasing during the middle of the year, particularly from June to September. This aligns with the 'Kiremt' or main rainy season in Ethiopia.

*   **Peak Rainy Months:** The months with the highest average total precipitation, indicating the peak of the rainy season, are typically July, August, and September.

*   **Dry Season:** The period from November to February appears to be the driest, with very low average precipitation, consistent with Ethiopia's dry season.

*   **Transition Months:** March, April, and May show a gradual increase in precipitation, suggesting a transition from the dry to the wet season. Similarly, October shows a decrease, marking the transition back to the dry season.

*   **Magnitude of Variation:** The difference between the dry and wet months is substantial, highlighting the pronounced seasonality of rainfall in the region. The peak months receive many times more precipitation than the driest months.

These patterns are typical for countries with a tropical monsoon climate, characterized by distinct wet and dry periods.

### Interpretation of Strongest Correlations

Based on the correlation heatmap (and general understanding of climate variables), here are three likely strong correlations:

1.  **T2M_MAX and T2M / T2M_MIN and T2M:** These are expected to be very highly positively correlated. Daily maximum (T2M_MAX) and minimum (T2M_MIN) temperatures are fundamental components of the average daily temperature (T2M). A higher maximum temperature will naturally lead to a higher average, and similarly for minimum temperature. These relationships demonstrate the internal consistency of the temperature measurements.

2.  **T2M and QV2M (Specific Humidity):** There is likely a strong positive correlation between average temperature (T2M) and specific humidity (QV2M). Warmer air can hold more moisture, so as temperatures increase, the amount of water vapor in the air (specific humidity) tends to increase as well. This is a basic principle of atmospheric physics.

3.  **T2M_RANGE and PS (Surface Pressure):** We might observe an inverse relationship between temperature range and surface pressure. Larger daily temperature ranges often occur during clear, dry conditions when there's less cloud cover to trap heat, allowing for greater diurnal temperature swings. These conditions are sometimes associated with higher pressure systems. However, this correlation might be less direct and influenced by other factors compared to temperature-humidity relationships.
