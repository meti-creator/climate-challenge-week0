```python
missing_values = df.isna().sum()
missing_percentage = (missing_values / len(df)) * 100

# Create a DataFrame for missing info
missing_info = pd.DataFrame({
    'Missing Values': missing_values,
    'Percentage (%)': missing_percentage
})

# Filter for columns with >5% missing values
high_missing = missing_info[missing_info['Percentage (%)'] > 5]

if not high_missing.empty:
    print("Columns with more than 5% missing values:")
    display(high_missing.sort_values(by='Percentage (%)', ascending=False))
    print("Note: High missing values in these columns could indicate data collection issues, sensor failures, or regional data gaps, potentially biasing analysis if not handled (e.g., via imputation or exclusion).")
else:
    print("No columns have more than 5% missing values.")
```

```python
from scipy.stats import zscore

z_score_cols = ['T2M', 'T2M_MAX', 'T2M_MIN', 'PRECTOTCORR', 'RH2M', 'WS2M', 'WS2M_MAX']

for col in z_score_cols:
    df[f'{col}_Zscore'] = zscore(df[col], nan_policy='omit')

# Identify outliers where |Z| > 3
outlier_mask = pd.DataFrame()
for col in z_score_cols:
    outlier_mask = pd.concat([outlier_mask, (df[f'{col}_Zscore'].abs() > 3)], axis=1)

df['is_outlier'] = outlier_mask.any(axis=1)

outlier_count = df['is_outlier'].sum()
print(f"Number of rows with at least one outlier (Z-score > 3): {outlier_count}")

display(df[df['is_outlier']].head())
```

## Outlier Treatment Strategy
**Decision:** Retain all identified outliers.

**Reasoning:**
*   **Scientific Relevance:** In climate datasets, outliers often represent genuine extreme weather events (e.g., peak temperatures or heavy precipitation). Removing them would result in a loss of critical information regarding climate volatility.
*   **Data Integrity:** There is no evidence to suggest these points are the result of sensor errors or manual data entry mistakes.
*   **Modeling Impact:** Retaining these points ensures the model accounts for "tail risks" and accurately reflects the real-world variance of the climate data.

```python
# The decision is to retain all outliers, so no explicit code for removal or capping is needed here.
# The 'is_outlier' column has already been added to the DataFrame to mark them.
# Display a sample of the DataFrame including the 'is_outlier' column to confirm.
display(df.head())
```

```python
# 1. Drop rows with missing values in more than 30% of the columns
limit = int(len(df.columns) * 0.7)
df = df.dropna(thresh=limit)

# 2. Apply forward-fill to the remaining missing weather variables
df = df.ffill()
df.head()
```

## Handling Missing Values
**Strategy:** Hybrid approach using row-wise deletion and Forward-Fill (ffill).

**1. Row Deletion (30% Threshold):**
Rows missing more than 30% of their features were removed. When a significant portion of
a record is missing, imputation becomes unreliable and may introduce synthetic noise
rather than capturing actual environmental trends.

**2. Forward-Fill (ffill) for Weather Variables:**
For the remaining missing values, I applied a forward-fill strategy.
* **Reasoning:** Climate data is chronological and highly autocorrelated. Weather
conditions at time $t$ are typically the most accurate predictors for time $t+1$.
Forward-filling preserves this temporal continuity without shifting the mean
as drastically as mean-imputation might.

# Time series Analysis

```python
import matplotlib.pyplot as plt


# Create a datetime index
if 'YEAR' in df.columns and 'MO' in df.columns:
    df['Date'] = pd.to_datetime(df[['YEAR', 'MO']].assign(DAY=1))
else:
    # Adjust this if your date column has a different name
    df['Date'] = pd.to_datetime(df['Date'])

df.set_index('Date', inplace=True)

# 2. Resample to Monthly Average
monthly_t2m = df['T2M'].resample('M').mean()

# 3. Identify Warmest and Coolest Months
warmest_month = monthly_t2m.idxmax()
warmest_val = monthly_t2m.max()

coolest_month = monthly_t2m.idxmin()
coolest_val = monthly_t2m.min()

# 4. Plotting
plt.figure(figsize=(14, 6))
plt.plot(monthly_t2m.index, monthly_t2m.values, color='teal', linewidth=2, label='Monthly Avg T2M')

# Annotate Warmest
plt.annotate(f'Warmest: {warmest_val:.1f}°C\n({warmest_month.strftime("%b %Y")})',
             xy=(warmest_month, warmest_val),
             xytext=(warmest_month, warmest_val + 1),
             arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=8),
             horizontalalignment='center', color='red', fontweight='bold')

# Annotate Coolest
plt.annotate(f'Coolest: {coolest_val:.1f}°C\n({coolest_month.strftime("%b %Y")})',
             xy=(coolest_month, coolest_val),
             xytext=(coolest_month, coolest_val - 2),
             arrowprops=dict(facecolor='blue', shrink=0.05, width=1, headwidth=8),
             horizontalalignment='center', color='blue', fontweight='bold')

plt.title('Monthly Average Temperature (T2M) in Ethiopia (2015–2026)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Temperature (°C)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()

plt.savefig('monthly_t2m_plot.png')
plt.show()
```

```python
monthly_prectotcorr = df['PRECTOTCORR'].resample('ME').sum()

# Identify peak rainy months
peak_rain_months_idx = monthly_prectotcorr.nlargest(3).index # Get top 3 rainy months
peak_rain_months_val = monthly_prectotcorr.nlargest(3).values

plt.figure(figsize=(14, 6))
monthly_prectotcorr.plot(kind='bar', color='skyblue', width=0.8)

plt.title('Monthly Total Precipitation (PRECTOTCORR) in Sudan (2015-2026)', fontsize=14)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Total Precipitation (mm)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(rotation=45, ha='right')

# Annotate peak rainy months
for i, month_idx in enumerate(peak_rain_months_idx):
    plt.annotate(f'{peak_rain_months_val[i]:.1f} mm\n({month_idx.strftime("%b %Y")})',
                 xy=(monthly_prectotcorr.index.get_loc(month_idx), peak_rain_months_val[i]),
                 xytext=(monthly_prectotcorr.index.get_loc(month_idx), peak_rain_months_val[i] + 5),
                 arrowprops=dict(facecolor='darkgreen', shrink=0.05, width=1, headwidth=8),
                 horizontalalignment='center', color='darkgreen', fontweight='bold')

plt.tight_layout()
plt.show()
```

## Trends and Anomalies in Climate Data

### Temperature (T2M)
*   **Seasonal Pattern:** The monthly average temperature plot shows a clear seasonal cycle, with temperatures generally increasing from the beginning of the year, peaking around mid-year (typically May-June), and then gradually decreasing towards the end of the year.
*   **Warmest and Coolest Months:** The warmest month observed is typically **May** or **June** with average temperatures exceeding 34°C, while the coolest months are usually around **January** or **February**, with averages below 25°C. This highlights the region's hot desert or semi-arid climate.
*   **Overall Stability:** Over the period from 2015 to 2026, there isn't an obvious strong long-term warming or cooling trend visible from the monthly averages, though inter-annual variability is present.

### Precipitation (PRECTOTCORR)
*   **Dry Season:** The bar chart for monthly total precipitation clearly indicates a prolonged dry season with minimal to no rainfall for many months of the year, particularly from October/November through April/May.
*   **Rainy Season:** There is a distinct rainy season, predominantly occurring in the **summer months** (e.g., June, July, August). These months show a significant spike in precipitation.
*   **Peak Rainy Months:** The annotated peak rainy months demonstrate that while the rainy season is concentrated, there can be substantial variation in the amount of rainfall received during these peak periods across different years.
*   **Extreme Events:** As noted in the initial descriptive statistics, the high standard deviation and maximum values for `PRECTOTCORR` (up to 66.49 mm in a single day, or over 200 mm in a month) suggest the occurrence of occasional, heavy rainfall events, which are characteristic of convective rainfall in arid and semi-arid regions. These events, while rare, contribute significantly to the total annual precipitation.

2.  **`T2M` and `QV2M`:** There is a strong positive correlation between `T2M` (Temperature at 2m) and `QV2M` (Specific Humidity at 2m). This suggests that warmer temperatures are associated with higher amounts of water vapor in the air. This makes sense as warmer air can hold more moisture.

3.  **`RH2M` and `PRECTOTCORR`:** A notable positive correlation exists between `RH2M` (Relative Humidity) and `PRECTOTCORR` (Precipitation). This is intuitive: higher relative humidity often precedes or accompanies precipitation events, as the air is saturated with moisture.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['PRECTOTCORR'], kde=True, bins=50)
plt.title('Distribution of PRECTOTCORR (Original Scale)', fontsize=14)
plt.xlabel('Precipitation (mm)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# Check for skewness and plot with log scale if heavily skewed
if df['PRECTOTCORR'].skew() > 1:
    plt.figure(figsize=(10, 6))
    sns.histplot(df['PRECTOTCORR'], kde=True, bins=50, log_scale=True)
    plt.title('Distribution of PRECTOTCORR (Log Scale)', fontsize=14)
    plt.xlabel('Precipitation (mm, Log Scale)', fontsize=12)
    plt.ylabel('Frequency', fontsize=12)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()
```

### Interpretation of PRECTOTCORR Distribution

The histogram for `PRECTOTCORR` (precipitation) clearly shows a highly skewed distribution. Most days have zero or very low precipitation, which is characteristic of arid or semi-arid regions. The distribution has a very long tail towards higher precipitation values, indicating rare but significant rainfall events.

When viewed on a log scale, the distribution becomes more discernible, revealing that while the majority of events are near zero, there's a spread of positive precipitation values, albeit at much lower frequencies. This emphasizes the episodic nature of rainfall in Sudan.
