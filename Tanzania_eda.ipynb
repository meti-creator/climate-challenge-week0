```python
missing_values = df.isna().sum()
missing_percentage = (missing_values / len(df)) * 100

# Create a DataFrame for missing info
missing_info = pd.DataFrame({
    'Missing Values': missing_values,
    'Percentage (%)': missing_percentage
})

# Filter for columns with >5% missing values
high_missing = missing_info[missing_info['Percentage (%)'] > 5]

if not high_missing.empty:
    print("Columns with more than 5% missing values:")
    display(high_missing.sort_values(by='Percentage (%)', ascending=False))
    print("Note: High missing values in these columns could indicate data collection issues, sensor failures, or regional data gaps, potentially biasing analysis if not handled (e.g., via imputation or exclusion).")
else:
    print("No columns have more than 5% missing values.")
```

```python
from scipy.stats import zscore

# Columns for Z-score calculation
z_score_cols = ['T2M', 'T2M_MAX', 'T2M_MIN', 'PRECTOTCORR', 'RH2M', 'WS2M', 'WS2M_MAX']

# Calculate Z-scores for the selected columns
for col in z_score_cols:
    df[f'{col}_Zscore'] = zscore(df[col], nan_policy='omit')

# Identify outliers where |Z| > 3
outlier_mask = pd.DataFrame()
for col in z_score_cols:
    outlier_mask = pd.concat([outlier_mask, (df[f'{col}_Zscore'].abs() > 3)], axis=1)

# A row is an outlier if any of the specified columns has an absolute Z-score > 3
df['is_outlier'] = outlier_mask.any(axis=1)

# Report the count of outliers
outlier_count = df['is_outlier'].sum()
print(f"Number of rows with at least one outlier (Z-score > 3): {outlier_count}")

display(df[df['is_outlier']].head())
```

## Outlier Treatment Strategy
**Decision:** Retain all 97 identified outliers.

**Reasoning:**
* **Scientific Relevance:** In climate datasets, outliers often represent genuine extreme weather events (e.g., peak temperatures or heavy precipitation). Removing them would result in a loss of critical information regarding climate volatility.
* **Data Integrity:** There is no evidence to suggest these points are the result of sensor errors or manual data entry mistakes.
* **Modeling Impact:** Retaining these points ensures the model accounts for "tail risks" and accurately reflects the real-world variance of the Ethiopian climate.

```python
# 1. Drop rows with missing values in more than 30% of the columns
limit = int(len(df.columns) * 0.7)
df = df.dropna(thresh=limit)

# 2. Apply forward-fill to the remaining missing weather variables
df = df.ffill()
df.head()
```

## Handling Missing Values
**Strategy:** Hybrid approach using row-wise deletion and Forward-Fill (ffill).

**1. Row Deletion (30% Threshold):**
Rows missing more than 30% of their features were removed. When a significant portion of
a record is missing, imputation becomes unreliable and may introduce synthetic noise
rather than capturing actual environmental trends.

**2. Forward-Fill (ffill) for Weather Variables:**
For the remaining missing values, I applied a forward-fill strategy.
* **Reasoning:** Climate data is chronological and highly autocorrelated. Weather
conditions at time $t$ are typically the most accurate predictors for time $t+1$.
Forward-filling preserves this temporal continuity without shifting the mean
as drastically as mean-imputation might.

# Time series Analysis

```python
import matplotlib.pyplot as plt


# Create a datetime index
if 'YEAR' in df.columns and 'MO' in df.columns:
    df['Date'] = pd.to_datetime(df[['YEAR', 'MO']].assign(DAY=1))
else:
    # Adjust this if your date column has a different name
    df['Date'] = pd.to_datetime(df['Date'])

df.set_index('Date', inplace=True)

# 2. Resample to Monthly Average
monthly_t2m = df['T2M'].resample('M').mean()

# 3. Identify Warmest and Coolest Months
warmest_month = monthly_t2m.idxmax()
warmest_val = monthly_t2m.max()

coolest_month = monthly_t2m.idxmin()
coolest_val = monthly_t2m.min()

# 4. Plotting
plt.figure(figsize=(14, 6))
plt.plot(monthly_t2m.index, monthly_t2m.values, color='teal', linewidth=2, label='Monthly Avg T2M')

# Annotate Warmest
plt.annotate(f'Warmest: {warmest_val:.1f}°C\n({warmest_month.strftime("%b %Y")})',
             xy=(warmest_month, warmest_val),
             xytext=(warmest_month, warmest_val + 1),
             arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=8),
             horizontalalignment='center', color='red', fontweight='bold')

# Annotate Coolest
plt.annotate(f'Coolest: {coolest_val:.1f}°C\n({coolest_month.strftime("%b %Y")})',
             xy=(coolest_month, coolest_val),
             xytext=(coolest_month, coolest_val - 2),
             arrowprops=dict(facecolor='blue', shrink=0.05, width=1, headwidth=8),
             horizontalalignment='center', color='blue', fontweight='bold')

plt.title('Monthly Average Temperature (T2M) in Ethiopia (2015–2026)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Temperature (°C)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()

plt.savefig('monthly_t2m_plot.png')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

monthly_prec = df['PRECTOTCORR'].resample('M').sum()

monthly_prec_df = monthly_prec.reset_index()
monthly_prec_df['Month'] = monthly_prec_df['Date'].dt.month_name()

avg_monthly_prec = monthly_prec_df.groupby('Month')['PRECTOTCORR'].mean().reindex([
    'January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'
])

peak_months = avg_monthly_prec.nlargest(3)

# Plotting
fig, ax = plt.subplots(figsize=(14, 7))
sns.barplot(x=avg_monthly_prec.index, y=avg_monthly_prec.values, palette='viridis', ax=ax)

ax.set_title('Average Monthly Total Precipitation (PRECTOTCORR) in Tanzania', fontsize=16)
ax.set_xlabel('Month', fontsize=12)
ax.set_ylabel('Average Total Precipitation (mm)', fontsize=12)
ax.tick_params(axis='x', rotation=45)
ax.grid(axis='y', linestyle='--', alpha=0.7)

# Annotate peak rainy season months
for month, value in peak_months.items():
    ax.annotate(f'{value:.1f}mm', xy=(month, value), xytext=(month, value + 5),
                ha='center', color='red', weight='bold', fontsize=10)

plt.tight_layout()
plt.show()
```

### Interpretation of Monthly Precipitation

**Key Observations from Average Monthly Total Precipitation (PRECTOTCORR) in Tanzania:**

1.  **Bimodal Rainfall Pattern**: The plot clearly shows two distinct rainy seasons throughout the year, which is characteristic of many equatorial and tropical regions.

2.  **First Rainy Season (Long Rains)**:
    *   This period extends primarily from **March to May**, with **April** standing out as the peak month for precipitation (approximately 298.1 mm). March and May also receive substantial rainfall (around 181.1 mm and 195.2 mm, respectively).

3.  **Second Rainy Season (Short Rains)**:
    *   A second, less intense rainy season occurs from **October to December**. November (approximately 155.4 mm) and December (around 145.3 mm) are the wettest months during this period.
    *   January also shows significant rainfall (around 142.6 mm), suggesting an extension of the 'short rains' or a transitional wet period.

4.  **Dry Season**: The months from **June to September** are notably dry, with July, August, and September recording very low average precipitation (all below 20 mm). February also acts as a relatively dry transitional month before the onset of the long rains.

These patterns are crucial for understanding the agricultural cycles, water resource management, and ecological dynamics in Tanzania.

# Distribution Analysis


```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Check for skewness
precipitation_skewness = df['PRECTOTCORR'].skew()

plt.figure(figsize=(10, 6))

if precipitation_skewness > 1: # A common threshold for high skewness
    # Apply log transformation to handle heavy skewness, adding a small constant to avoid log(0)
    sns.histplot(np.log1p(df['PRECTOTCORR']), kde=True, color='skyblue')
    plt.title(f'Distribution of Log-transformed Precipitation (PRECTOTCORR)\nSkewness: {precipitation_skewness:.2f}', fontsize=16)
    plt.xlabel('Log(PRECTOTCORR + 1)', fontsize=12)
else:
    sns.histplot(df['PRECTOTCORR'], kde=True, color='skyblue')
    plt.title(f'Distribution of Precipitation (PRECTOTCORR)\nSkewness: {precipitation_skewness:.2f}', fontsize=16)
    plt.xlabel('PRECTOTCORR (mm)', fontsize=12)

plt.ylabel('Frequency', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Interpretation of Precipitation Distribution (PRECTOTCORR)

Precipitation data (PRECTOTCORR) is typically right-skewed, meaning there are many days with low or no rainfall and fewer days with very high rainfall events. If a log transformation was applied, it indicates significant positive skewness, making the original data heavily concentrated at lower values with a long tail towards higher values. The log transformation helps in normalizing the distribution, making it more amenable for statistical analysis that assumes normality.

*   **Shape:** If log-transformed, the distribution should appear more symmetrical or bell-shaped, indicating that the transformation successfully reduced the skewness.
*   **Concentration:** The bulk of the precipitation events, even after transformation, will likely show a concentration around lower to moderate values, corresponding to light to average rainfall days.
*   **Outliers/Extremes:** The right tail, even in the transformed data, represents rarer, heavy rainfall events, which are crucial for understanding extreme weather patterns.
